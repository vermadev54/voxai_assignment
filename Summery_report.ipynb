# 📊 Contact Center Analytics Summary Report

## 📌 Key Insights

- **IVR Navigation**: Most common path was `Start>Billing`, suggesting a high need for billing-related human assistance.
- **Call Drop-offs**: Higher drop-off observed at early stages of IVR flow (`Start>TechSupport`), indicating user frustration or confusion.
- **Agent Performance**:
  - Wide variability in **Average Handling Time (AHT)** and **First Call Resolution (FCR)** across agents.
  - Top-performing agents demonstrated higher FCR and lower escalation rates.
- **Customer Sentiment**:
  - 40% positive, 38% neutral, and 22% negative sentiment overall.
  - Sentiment worsens in calls that end with transfers or escalations.
- **Bot Failures**:
  - `no_match` and `no_input` were leading bot failure reasons.
  - Majority of live agent transfers originated from failed self-service attempts.

## 🔬 Methodology

- **Synthetic Data Generation**: Simulated 2,000 calls with realistic features (IVR paths, durations, agents, customer utterances, sentiments).
- **Data Preprocessing**: Handled missing values, standardized formats, and anonymized sensitive identifiers.
- **EDA**: Visualized IVR paths, drop-offs, AHT, FCR, and transfer behavior.
- **NLP & ML**:
  - Used keyword extraction and clustering for **intent analysis**.
  - Applied sentiment analysis on customer utterances to track emotional trends.
- **Visualization**: Built an interactive Streamlit dashboard with filters for agent, team, and date.

## 🤖 Decisions & Justifications

- **Synthetic Data** was crafted to mirror realistic contact center conditions and include all needed labels for analytics and ML modeling.
- **Evaluation Metrics**:
  - **Accuracy** for bot success/failure classification.
  - **F1 Score** for imbalanced intent classification.
  - For LLM-generated insights: manual validation + relevance ranking.
- **Tooling Choices**:
  - **Python/Pandas** for analysis.
  - **Streamlit** for real-time dashboarding.
  - **Scikit-learn/NLTK** for NLP tasks.

## ✅ Recommendations

1. **Improve IVR Routing**:
   - Re-design unclear paths like `Start>TechSupport` to reduce drop-offs.
   - Introduce early clarification prompts.

2. **Bot Optimization**:
   - Retrain NLU models with failed query patterns (`no_match`, `ambiguous_query`).
   - Increase fallback confidence threshold or route sooner to human agents.

3. **Agent Training**:
   - Focus on agents with high escalations and low FCR.
   - Use sentiment trends as coaching feedback.

4. **Intent Analytics**:
   - Use detected topics for proactive FAQ updates and self-service enhancements.

5. **Monitoring & Scaling**:
   - Deploy dashboards with real-time KPIs segmented by team, shift, and customer type.
